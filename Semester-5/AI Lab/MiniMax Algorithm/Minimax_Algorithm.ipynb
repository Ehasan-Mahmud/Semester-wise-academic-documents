```python
game_tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['-1', '4'],
    'E': ['2', '6'],
    'F': ['-3', '-5'],
    'G': ['0', '7'],
    '-1': [],
    '4': [],
    '2': [],
    '6': [],
    '-3': [],
    '-5': [],
    '0': [],
    '7': []
}

heuristic_values = {
    '-1': -1,
    '4': 4,
    '2': 2,
    '6': 6,
    '-3': -3,
    '-5': -5,
    '0': 0,
    '7': 7
}

def evaluate(node):
    return heuristic_values.get(node, 0)

def get_children(node):
    return game_tree.get(node, [])

def minimax(node, is_maximizing_player):

    if not get_children(node):
        return evaluate(node), [node]

    if is_maximizing_player:
        max_eval = float('-inf')
        best_path = []
        for child in get_children(node):
            eval, path = minimax(child, False)
            if eval > max_eval:
                max_eval = eval
                best_path = [node] + path
        return max_eval, best_path
    else:
        min_eval = float('inf')
        best_path = []
        for child in get_children(node):
            eval, path = minimax(child, True)
            if eval < min_eval:
                min_eval = eval
                best_path = [node] + path
        return min_eval, best_path

score, path = minimax('A', is_maximizing_player=True)

print("Best achievable score for Max:", score)
print("Best path:", " -> ".join(path))
```
