```python
import heapq
def astar_route_search(city_map, heuristics, start, goal):
    open_list = []
    heapq.heappush(open_list, (heuristics[start], 0, start, [start]))
    visited = set()
    g_costs = {city: float('inf') for city in city_map}
    g_costs[start] = 0
    move_count = 1
    while open_list:
        f_score, g_score, current, path = heapq.heappop(open_list)
        if current in visited:
            continue
        visited.add(current)
        print(f"\nStep {move_count}:")
        print(f"Current city: {current}")
        print(f"Path so far: {' - '.join(path)}")
        print(f"g({current}) = {g_score}")
        print(f"h({current}) = {heuristics[current]}")
        print(f"f({current}) = {f_score}")
        print("Frontier:", [city for (_, _, city, _) in open_list])
        if current == goal:
            return path, g_score
        for neighbor, cost in city_map[current].items():
            if neighbor in visited:
                continue
            tentative_g = g_score + cost
            if tentative_g < g_costs[neighbor]:
                g_costs[neighbor] = tentative_g
                f_neighbor = tentative_g + heuristics[neighbor]
                heapq.heappush(open_list, (f_neighbor, tentative_g, neighbor, path + [neighbor]))
        move_count += 1
    return None, None
romanian_map = {
    'Arad': {'Zerind': 75, 'Sibiu': 140, 'Timisoara': 118},
    'Zerind': {'Arad': 75, 'Oradea': 71},
    'Oradea': {'Zerind': 71, 'Sibiu': 151},
    'Sibiu': {'Arad': 140, 'Oradea': 151, 'Fagaras': 99, 'Rimnicu Vileea': 80},
    'Timisoara': {'Arad': 118, 'Lugoj': 111},
    'Lugoj': {'Timisoara': 111, 'Mehadia': 70},
    'Mehadia': {'Lugoj': 70, 'Drobeta': 75},
    'Drobeta': {'Mehadia': 75, 'Craiova': 120},
    'Craiova': {'Drobeta': 120, 'Rimnicu Vileea': 146, 'Pitesti': 138},
    'Rimnicu Vileea': {'Sibiu': 80, 'Craiova': 146, 'Pitesti': 97},
    'Fagaras': {'Sibiu': 99, 'Bucharest': 211},
    'Pitesti': {'Rimnicu Vileea': 97, 'Craiova': 138, 'Bucharest': 101},
    'Bucharest': {'Fagaras': 211, 'Pitesti': 101, 'Giurgiu': 90, 'Urziceni': 85},
    'Giurgiu': {'Bucharest': 90},
    'Urziceni': {'Bucharest': 85, 'Hirsova': 98, 'Vaslui': 142},
    'Hirsova': {'Urziceni': 98, 'Eforie': 86},
    'Eforie': {'Hirsova': 86},
    'Vaslui': {'Urziceni': 142, 'Iasi': 92},
    'Iasi': {'Vaslui': 92, 'Neamt': 87},
    'Neamt': {'Iasi': 87}
}
heuristics = {
    'Arad': 366,
    'Bucharest': 0,
    'Craiova': 160,
    'Drobeta': 242,
    'Eforie': 161,
    'Fagaras': 176,
    'Giurgiu': 77,
    'Hirsova': 151,
    'Iasi': 226,
    'Lugoj': 244,
    'Mehadia': 241,
    'Neamt': 234,
    'Oradea': 380,
    'Pitesti': 100,
    'Rimnicu Vileea': 193,
    'Sibiu': 253,
    'Timisoara': 329,
    'Urziceni': 80,
    'Vaslui': 199,
    'Zerind': 374
}
print("A* Search: Finding the most efficient path from Arad to Bucharest")
path, total_cost = astar_route_search(romanian_map, heuristics, 'Arad', 'Bucharest')
if path:
    print("\nFinal Path:", " - ".join(path))
    print("Total Cost:", total_cost)
else:
    print("No path found.")
```
